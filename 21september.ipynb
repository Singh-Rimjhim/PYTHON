```python
import numpy as np
import pandas as pd
```

```python
#storing data in tabular format
data = {
    'Name': ['Rimjhim', 'Jhalak', 'Ritu'],
    'Age': [19, 18, 20],
    'City': ['Jaunpur', 'Aswan', 'Gorakhpur']
}

df = pd.DataFrame(data)
print(df)
```

```python
#only works on numeric columns (like Age).
print(df.describe())
```

```python
#Structure of DataFrame (columns, dtypes, memory)
print(df.info())
```

```python
#First 5 rows := df(head)
#df.tail() := Last 5 rows
#Column-wise median => df.mean()
#Column-wise mean => df.median()
#Frequency of values in a column
df.value_counts()	
```
