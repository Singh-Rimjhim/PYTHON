## Violinplot
A violin plot plays a similar role as a box-and-whisker plot. It shows the distribution of data points after grouping by one (or more) variables. Unlike a box plot, each violin is drawn using a kernel density estimate of the underlying distribution.

```python
sns.violinplot(fd, palette='rainbow')
```

| **Bar Graph** | Compare categories | Which product has the highest sales? |
| **Histogram** | Understand distribution of numerical data | How are students' marks distributed? |
| **Boxplot** | Find spread, median & outliers | Are there unusual salary values? |
| **Pie Chart** | Show parts of a whole | What percentage of sales comes from each category? |
| **Scatter Plot** | Relationship between two numerical variables | Does study time relate to marks? |
| **Line Chart** | Trends over time | How did sales change over months? |
| **Count Plot** | Count observations in categories | How many students are in each branch? |
| **Heatmap** | Show values/intensity in a matrix | Which variables are highly correlated? |
| **Violin Plot** | Distribution + density + comparison | How does salary distribution differ by department? |
| **Pairplot** | Relationships among multiple numerical variables | How are all features related to each other? |
| **Jointplot** | Relationship between two variables + their distributions | How are height and weight related? |
| **Area Chart** | Show magnitude/change over time | How has cumulative revenue changed? |
