```python
data = [2, 3, 4, 5, 6, 7, 8]
plt.hist(data, bins = 5, color='blue', edgecolor='black', zorder=2)
#use grid at bg (zorder=1) so that the grid is behind the bars
plt.grid(True, zorder=1)
plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
import seaborn as sns
data = [2, 3, 4, 5, 6, 7, 8]
plt.hist(data, bins=5, color='blue', edgecolor='black', alpha=0.6, density=True, zorder=2)
sns.kdeplot(data, color='blue', linewidth=2, zorder=3)
plt.grid(True, zorder=1)
plt.title('Histogram with KDE')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
```
