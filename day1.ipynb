```python
# 6) Mutable vs Immutable
def mutable_vs_immutable():
    # Immutable example (string)
    s = "hello"
    print("Original string:", s)
    s = s.replace("h", "H")
    print("Modified string:", s)

    # Mutable example (list)
    lst = [1, 2, 3]
    print("Original list:", lst)
    lst[0] = 99
    print("Modified list:", lst)
mutable_vs_immutable()
```

```python
# 7) Global variable modification
x = 10

def modify_global():
    global x
    x = x + 5
    print("Inside function, x =", x)

modify_global()
print("Outside function, x =", x)
```

```python

# 8) Using del
def delete_variable():
    y = 100
    print("Before deletion, y =", y)
    del y
    try:
        print(y)
    except NameError:
        print("y is deleted and no longer exists.")

delete_variable()
```

```python
# 9) Convert input strings to integers
def string_to_int_arithmetic():
    a = "12"
    b = "8"
    num1 = int(a)
    num2 = int(b)
    print("Sum:", num1 + num2)
    print("Product:", num1 * num2)

string_to_int_arithmetic()
```

```python
# 10) Shallow vs Deep Copy
import copy

def shallow_deep_copy():
    original = [[1, 2], [3, 4]]

    # Shallow copy
    shallow = copy.copy(original)
    shallow[0][0] = 99
    print("Original after shallow change:", original)

    # Deep copy
    deep = copy.deepcopy(original)
    deep[1][1] = 77
    print("Original after deep change:", original)
    print("Deep copy:", deep)

shallow_deep_copy()
```
